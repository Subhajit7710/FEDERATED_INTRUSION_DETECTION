```python
# Cell 10 — Federated learning configuration

NUM_CLIENTS = 4
NUM_ROUNDS = 3
LOCAL_EPOCHS = 1

BATCH_SIZE = 1024
LEARNING_RATE = 0.001

print("Clients:", NUM_CLIENTS)
print("Rounds:", NUM_ROUNDS)
print("Local epochs:", LOCAL_EPOCHS)
print("Batch size:", BATCH_SIZE)
print("Learning rate:", LEARNING_RATE)
print("Device:", DEVICE)
```

```python
# Cell 11 — Federated client dataset

import numpy as np

from torch.utils.data import Dataset, DataLoader


class ClientDataset(Dataset):

    def __init__(self, X, y, indices):
        self.X = X
        self.y = y
        self.indices = indices

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, index):

        real_index = self.indices[index]

        x = np.array(
            self.X[real_index],
            dtype=np.float32,
            copy=True
        )

        y = np.float32(
            self.y[real_index]
        )

        return (
            torch.from_numpy(x),
            torch.tensor(y, dtype=torch.float32)
        )


def load_client_data(client_id):

    indices = np.load(
        CLIENT_DIR / f"client_{client_id}_indices.npy"
    )

    X_train = np.load(
        DATA_DIR / "X_train.npy",
        mmap_mode="r"
    )

    y_train = np.load(
        DATA_DIR / "y_train.npy",
        mmap_mode="r"
    )

    dataset = ClientDataset(
        X_train,
        y_train,
        indices
    )

    loader = DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=0,
        pin_memory=(DEVICE.type == "cuda")
    )

    return loader, len(dataset)

print("Client loader ready.")
```

```python
# Cell 12 — Local model training

import torch.nn as nn


def train_local_model(model, train_loader, epochs=1):

    model.to(DEVICE)
    model.train()

    criterion = nn.BCEWithLogitsLoss()

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LEARNING_RATE
    )

    total_loss = 0.0

    for epoch in range(epochs):

        epoch_loss = 0.0

        for X_batch, y_batch in train_loader:

            X_batch = X_batch.to(
                DEVICE,
                non_blocking=True
            )

            y_batch = y_batch.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad()

            logits = model(
                X_batch
            ).squeeze(1)

            loss = criterion(
                logits,
                y_batch
            )

            loss.backward()
            optimizer.step()

            epoch_loss += loss.item()

        total_loss = (
            epoch_loss /
            len(train_loader)
        )

        print(
            f"Epoch {epoch + 1}/{epochs} - "
            f"Loss: {total_loss:.4f}"
        )

    return total_loss
```

```python
# Cell 13 — GPU training smoke test

client_loader, client_size = load_client_data(1)

print("Client 1 samples:", client_size)
print("Batches:", len(client_loader))

smoke_model = IDSModel(
    input_size=78
)

smoke_loss = train_local_model(
    smoke_model,
    client_loader,
    epochs=1
)

print("Smoke test completed.")
print("Final loss:", smoke_loss)
```

```python
# Cell 15 — Flower client

class FlowerIDSClient(fl.client.NumPyClient):

    def __init__(self, client_id):

        self.client_id = client_id

        self.train_loader, self.num_samples = (
            load_client_data(client_id)
        )

        self.model = IDSModel(
            input_size=78
        ).to(DEVICE)

    def get_parameters(self, config):

        return [
            value.detach().cpu().numpy()
            for value in self.model.state_dict().values()
        ]

    def set_parameters(self, parameters):

        state_dict = self.model.state_dict()

        new_state_dict = {
            key: torch.tensor(
                value,
                dtype=state_dict[key].dtype
            )
            for key, value in zip(
                state_dict.keys(),
                parameters
            )
        }

        self.model.load_state_dict(
            new_state_dict
        )

    def fit(self, parameters, config):

        self.set_parameters(parameters)

        local_epochs = int(
            config.get(
                "local_epochs",
                LOCAL_EPOCHS
            )
        )

        loss = train_local_model(
            self.model,
            self.train_loader,
            epochs=local_epochs
        )

        return (
            self.get_parameters(config),
            self.num_samples,
            {"loss": float(loss)}
        )


def client_fn(context):

    client_id = int(
        context.node_config["partition-id"]
    ) + 1

    return FlowerIDSClient(
        client_id
    ).to_client()


print("Flower client ready.")
```

```python
# File: Colab notebook Cell 20 — FedAvg strategy

import numpy as np
import torch
import flwr as fl

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Load validation data
X_val = np.load(
    DATA_DIR / "X_val.npy",
    mmap_mode="r"
)

y_val = np.load(
    DATA_DIR / "y_val.npy",
    mmap_mode="r"
)


def evaluate_global_model(
    server_round,
    parameters,
    config
):
    model = IDSModel(input_size=78)

    state_dict = model.state_dict()

    new_state_dict = {
        key: torch.tensor(
            value,
            dtype=state_dict[key].dtype
        )
        for key, value in zip(
            state_dict.keys(),
            parameters
        )
    }

    model.load_state_dict(new_state_dict)
    model.to(DEVICE)
    model.eval()

    predictions = []
    actual = []

    batch_size = 4096

    with torch.no_grad():

        for start in range(0, len(X_val), batch_size):

            end = min(
                start + batch_size,
                len(X_val)
            )

            X_batch = np.array(
                X_val[start:end],
                dtype=np.float32,
                copy=True
            )

            y_batch = np.array(
                y_val[start:end],
                dtype=np.float32,
                copy=True
            )

            X_tensor = torch.from_numpy(
                X_batch
            ).to(
                DEVICE,
                non_blocking=True
            )

            logits = model(
                X_tensor
            ).squeeze(1)

            probs = torch.sigmoid(logits)

            preds = (
                probs >= 0.5
            ).cpu().numpy().astype(np.int32)

            predictions.extend(preds)

            actual.extend(
                y_batch.astype(np.int32)
            )

    accuracy = accuracy_score(
        actual,
        predictions
    )

    precision = precision_score(
        actual,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        actual,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        actual,
        predictions,
        zero_division=0
    )

    print(
        f"Round {server_round} - "
        f"Accuracy: {accuracy:.4f}, "
        f"Precision: {precision:.4f}, "
        f"Recall: {recall:.4f}, "
        f"F1: {f1:.4f}"
    )

    return 0.0, {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }


# Store final aggregated parameters
final_parameters = None


class SaveFinalParametersStrategy(
    fl.server.strategy.FedAvg
):

    def aggregate_fit(
        self,
        server_round,
        results,
        failures
    ):
        global final_parameters

        aggregated_parameters, aggregated_metrics = (
            super().aggregate_fit(
                server_round,
                results,
                failures
            )
        )

        if aggregated_parameters is not None:
            final_parameters = aggregated_parameters

        return (
            aggregated_parameters,
            aggregated_metrics
        )


# Create FedAvg strategy
strategy = SaveFinalParametersStrategy(
    fraction_fit=1.0,
    fraction_evaluate=0.0,
    min_fit_clients=4,
    min_available_clients=4,
    evaluate_fn=evaluate_global_model
)

print(
    "FedAvg strategy with final model capture "
    "configured successfully."
)
```

```python
# Cell 17 — Federated simulation

print("Starting federated simulation...")
print("Clients:", NUM_CLIENTS)
print("Rounds:", NUM_ROUNDS)
print("Device:", DEVICE)

history = fl.simulation.start_simulation(
    client_fn=client_fn,
    num_clients=NUM_CLIENTS,
    config=fl.server.ServerConfig(
        num_rounds=NUM_ROUNDS
    ),
    strategy=strategy,
    client_resources={
        "num_cpus": 1,
        "num_gpus": 1
    }
)

print("Federated simulation completed.")
```
